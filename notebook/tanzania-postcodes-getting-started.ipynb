# Tanzania Postcodes: Getting Started

This notebook shows how to load the dataset and answer common questions: the postcode for a ward, the wards in a district, and where a street is.

Each row is one street or village, with its region, district, ward and the ward's 5-digit postcode (`ward_code`).

```python
import glob

import matplotlib.pyplot as plt
import pandas as pd

path = glob.glob("/kaggle/input/**/all_tanzania.csv", recursive=True)[0]
# Read every column as text so postcodes keep their exact digits
df = pd.read_csv(path, dtype=str)
df.head()
```

## Overview

```python
pd.Series({
    "regions": df.region.nunique(),
    "districts": df[["region", "district"]].drop_duplicates().shape[0],
    "wards (postcodes)": df.ward_code.nunique(),
    "streets / villages": len(df),
})
```

## Wards per region

```python
wards = df.groupby("region").ward_code.nunique().sort_values()

ax = wards.plot.barh(figsize=(8, 9), color="#1EB53A")
ax.set_xlabel("Number of wards (postcodes)")
ax.set_ylabel("")
ax.set_title("Wards per region")
plt.tight_layout()
plt.show()
```

## Find the postcode for a ward

```python
def postcode(ward):
    match = df[df.ward.str.upper() == ward.upper()]
    return match[["region", "district", "ward", "ward_code"]].drop_duplicates()

postcode("Kariakoo")
```

## List the wards in a district

```python
def wards_in(district):
    match = df[df.district.str.upper() == district.upper()]
    return match[["ward", "ward_code"]].drop_duplicates().sort_values("ward_code").reset_index(drop=True)

wards_in("Arusha CBD")
```

## Search for a street or village

```python
def find_street(name):
    match = df[df.street.str.contains(name, case=False, na=False)]
    return match[["street", "ward", "ward_code", "district", "region"]].drop_duplicates()

find_street("Sanawari")
```
